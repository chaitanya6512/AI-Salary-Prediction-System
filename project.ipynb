```python
skill_salary = {}

# Loop through every job posting
for _, row in df.iterrows():

    skills = row["skills"]
    salary = row["salary_avg"]

    # Only process if skills is a list
    if isinstance(skills, list):

        for skill in skills:

            skill = skill.lower().strip()

            if skill not in skill_salary:
                skill_salary[skill] = []

            skill_salary[skill].append(salary)

# Calculate average salary for each skill
average_skill_salary = {}

for skill, salaries in skill_salary.items():
    average_skill_salary[skill] = sum(salaries) / len(salaries)

# Convert to DataFrame
skill_salary_df = (
    pd.DataFrame(
        average_skill_salary.items(),
        columns=["Skill", "Average Salary"]
    )
)

# Sort by highest salary
skill_salary_df = skill_salary_df.sort_values(
    by="Average Salary",
    ascending=False
)

# Show Top 20 highest-paying skills
skill_salary_df.head(20)
```

```python
# ==========================================================
# PHASE 2
# STEP 2.5 - DATE FEATURE ENGINEERING
# ==========================================================

import re
import numpy as np

def convert_post_days(text):

    if pd.isna(text):
        return np.nan

    text = text.lower().strip()

    if text == "a day ago":
        return 1

    elif text == "a week ago":
        return 7

    elif text == "a month ago":
        return 30

    elif text == "today":
        return 0

    number = re.findall(r"\d+", text)

    if number:

        number = int(number[0])

        if "day" in text:
            return number

        elif "week" in text:
            return number * 7

        elif "month" in text:
            return number * 30

        elif "year" in text:
            return number * 365

    return np.nan


df["days_since_posted"] = df["post_date"].apply(convert_post_days)

print("✅ Date Feature Created Successfully!")
```

```python
# ==========================================================
# PHASE 2
# STEP 2.6 - POSTING CATEGORY
# ==========================================================

def posting_category(days):

    if pd.isna(days):
        return "Unknown"

    elif days <= 7:
        return "Very Recent"

    elif days <= 30:
        return "Recent"

    elif days <= 90:
        return "Moderate"

    else:
        return "Old"


df["posting_category"] = df["days_since_posted"].apply(posting_category)

print("✅ Posting Category Created Successfully!")
```

```python
# ==========================================================
# PHASE 2
# STEP 2.7 - STATE & COUNTRY FEATURE ENGINEERING
# ==========================================================

import pandas as pd

# ---------- Country Extraction ----------

def extract_country(location):

    if pd.isna(location):
        return "Unknown"

    location = location.lower()

    if "canada" in location:
        return "Canada"

    elif "india" in location:
        return "India"

    elif "switzerland" in location:
        return "Switzerland"

    elif "germany" in location:
        return "Germany"

    elif "united kingdom" in location:
        return "United Kingdom"

    elif "uk" in location:
        return "United Kingdom"

    elif "france" in location:
        return "France"

    elif "australia" in location:
        return "Australia"

    elif "singapore" in location:
        return "Singapore"

    elif "netherlands" in location:
        return "Netherlands"

    else:
        return "USA"


# ---------- USA or International ----------

def is_usa(country):
    return 1 if country == "USA" else 0


def is_international(country):
    return 0 if country == "USA" else 1


# ---------- Apply Features ----------

df["country"] = df["location"].apply(extract_country)

df["is_USA"] = df["country"].apply(is_usa)

df["is_International"] = df["country"].apply(is_international)

print("✅ Country Features Created Successfully!")
```

```python
# ==========================================================
# PHASE 3
# STEP 3.9 - IMPORT ALL LIBRARIES
# ==========================================================

# -----------------------------
# Numerical Computing
# -----------------------------
import numpy as np
import pandas as pd

# -----------------------------
# Visualization
# -----------------------------
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# Train Test Split
# -----------------------------
from sklearn.model_selection import (
    train_test_split,
    GridSearchCV,
    RandomizedSearchCV
)

# -----------------------------
# Machine Learning Models
# -----------------------------
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

# XGBoost
from xgboost import XGBRegressor

# CatBoost
from catboost import CatBoostRegressor

# LightGBM
from lightgbm import LGBMRegressor

# -----------------------------
# Evaluation Metrics
# -----------------------------
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# -----------------------------
# Model Saving
# -----------------------------
import joblib

# -----------------------------
# Ignore Warnings
# -----------------------------
import warnings
warnings.filterwarnings("ignore")

print("="*60)
print("✅ ALL LIBRARIES IMPORTED SUCCESSFULLY")
print("="*60)
```

```python
# ==========================================================
# PHASE 3
# STEP 3.10 - TRAIN ALL MODELS
# ==========================================================

models = {

    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        random_state=42,
        n_estimators=100,
        n_jobs=-1
    ),

    "XGBoost": XGBRegressor(
        random_state=42,
        n_estimators=100,
        verbosity=0
    ),

    "CatBoost": CatBoostRegressor(
        random_state=42,
        verbose=0
    ),

    "LightGBM": LGBMRegressor(
        random_state=42
    )

}

results = []

for name, model in models.items():

    print("="*60)
    print(f"Training {name}...")
    print("="*60)

    # Train
    model.fit(X_train, y_train)

    # Prediction
    prediction = model.predict(X_test)

    # Metrics
    mae = mean_absolute_error(y_test, prediction)

    rmse = np.sqrt(
        mean_squared_error(y_test, prediction)
    )

    r2 = r2_score(y_test, prediction)

    results.append({

        "Model": name,
        "MAE": round(mae,2),
        "RMSE": round(rmse,2),
        "R2 Score": round(r2,4)

    })

print("\n✅ All Models Trained Successfully!")
```

```python
# ==========================================================
# MODEL COMPARISON
# ==========================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="R2 Score",
    ascending=False
)

results_df.reset_index(drop=True, inplace=True)

print(results_df)
```

```python
best_model = results_df.iloc[0]

print("="*60)
print("🏆 BEST MODEL")
print("="*60)

print(best_model)
```

```python
# ==========================================================
# PHASE 4
# STEP 4.1 - XGBOOST PARAMETER GRID
# ==========================================================

param_grid = {

    "n_estimators": [100, 200, 300],

    "max_depth": [3, 5, 7],

    "learning_rate": [0.01, 0.05, 0.1],

    "subsample": [0.8, 1.0],

    "colsample_bytree": [0.8, 1.0]

}

print("Total Combinations:",
      3*3*3*2*2)
```
